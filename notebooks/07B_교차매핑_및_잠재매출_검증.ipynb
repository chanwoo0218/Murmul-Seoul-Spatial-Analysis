상위 잠재 블록의 후속성과 중첩과 공간적 군집성을 추가로 검증합니다.


```python
import pandas as pd
def topk_iou(y_true, y_pred, ratio=0.1):
    k = max(1, int(len(y_true)*ratio))
    a = set(y_true.nlargest(k).index)
    b = set(y_pred.nlargest(k).index)
    return len(a & b) / len(a | b)
# df = pd.read_csv('../outputs/validation_2020_2022.csv').dropna()
# print('IoU@10%', topk_iou(df['actual_sales_2022'], df['potential_sales_2020']))
# Moran's I는 좌표계가 통일된 공간가중치 행렬로 별도 계산합니다.
```
