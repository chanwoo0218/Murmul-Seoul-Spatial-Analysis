# Dasymetric Mapping
상권 매출을 블록별 상업활동 점수와 교차면적을 이용해 재배분합니다.


```python
import pandas as pd
import numpy as np
def allocate_sales(df, district_col='TRDAR_CD', sales_col='sales'):
    out = df.copy()
    scaled = ['walk_access_scaled','bus_access_scaled','subway_access_scaled','road_access_scaled','poi_density_scaled','building_capacity_scaled']
    out['commercial_score'] = out[scaled].mean(axis=1)
    out['weight_raw'] = out['commercial_score'] * out['overlap_ratio']
    denom = out.groupby(district_col)['weight_raw'].transform('sum')
    out['allocation_weight'] = np.where(denom>0, out['weight_raw']/denom, 0)
    out['potential_sales'] = out[sales_col] * out['allocation_weight']
    return out
```
