# 역세권 블록 검증
역 위치와 블록 공간자료의 중첩·거리·좌표계를 점검하는 노트북입니다.


```python
from pathlib import Path
import geopandas as gpd
import pandas as pd
stations = pd.read_csv('../data/public/서울교통공사_역사좌표.csv')
stations_gdf = gpd.GeoDataFrame(stations, geometry=gpd.points_from_xy(stations['경도'], stations['위도']), crs=4326).to_crs(5179)
display(stations_gdf.head())
```
