# 기후동행카드 효과 분석
ZIP의 정책·승하차 분석 노트북을 상대경로 기준으로 정리한 공개용 사본입니다.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA = Path('../data/processed')
```

```python
monthly = pd.read_csv(DATA / '기후동행카드_효과분석_기초결과__월별_승하차_집계.csv')
comparison = pd.read_csv(DATA / '기후동행카드_효과분석_기초결과__전년대비_승하차_비교.csv')
display(monthly.head())
display(comparison.head())
```

```python
monthly.plot(x=monthly.columns[0], figsize=(10,4))
plt.title('월별 승하차 추이')
plt.tight_layout(); plt.show()
```
